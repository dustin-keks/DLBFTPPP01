# Vergleich von Aktien
## 1. Datenbeschaffung
Aktien und Zeitraum auswählen.

```python
import yfinance

tickers = {
    "AAPL": "Apple (USA)",
    "NVDA": "NVIDIA (USA)",
    "ASML": "ASML (Niederlande)",
    "SAP": "SAP (Deutschland)",
    "TM": "Toyota (Japan)",
    "SONY": "Sony (Japan)",
}

period = "5y"
```

Daten herunterladen und anzeigen.

```python
data = yfinance.download(list(tickers.keys()), period=period, auto_adjust=True)
data.head()
```

Sanity-Check

```python
print("Zeitraum von", data.index.min().date(), "bis", data.index.max().date())
print("Anzahl der Handelstage:", len(data))
print("Fehlende Werte je Aktie:")
print(data.isna().sum())
```
